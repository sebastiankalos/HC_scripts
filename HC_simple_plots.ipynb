```python
import numpy as np
import matplotlib.pyplot as plt

# ---------- Panel 1: your time-domain sketch (keep style) ----------
# Data
t = np.linspace(-4, 4, 1000)   # this is t/τ for the sketch panel
A = 1
tau = 1
def get_omega_t(A,t):
    omega_t = A*t*np.exp(-t**2/(2*tau**2))
    return omega_t

omega_t = get_omega_t(A,t)
omega_t = omega_t / np.max(omega_t)/2  # normalize

# Create figure with subplots (no shared x)
fig, ax = plt.subplots(1, 2, figsize=(15, 4), sharex=False)

# First subplot: original function (keep your look)
ax[0].plot(t, np.exp(-t**2 / (2*tau**2)), color="black", alpha=0.7, label=r'$I(t)$')
ax[0].set_ylim(0,1.1)
ax[0].set_ylabel(r"Normalized Temporal Intensity (a. u.)", color="black", size=10)
ax[0].text(s=r'$I(t)$', x=-1.35, y=0.75, size=13)

# Second y-axis on first subplot
ax0 = ax[0].twinx()
ax0.plot(t, omega_t, color="green", label=r"$\omega(t)$")
ax0.set_ylim(-1,1)
ax0.axhline(y=0, ls='dotted', color='green', lw=1)
ax0.set_ylabel(r"$\omega(t)-\omega_0$  (a.u.)", color="green", size=11)
ax0.set_yticks([0])
ax0.tick_params(axis='y', colors='green')
ax0.vlines(x=-1, ymin=-1, ymax=-0.5, color='green', ls='dotted', lw=1)
ax0.vlines(x=1,  ymin=-1, ymax=0.5,  color='green', ls='dotted', lw=1)

# Improve layout (keep your labels/annotations)
ax[0].set_xlabel(r'$t/\tau$', size=12)
ax0.text(s=r'$\Delta \omega(t)$', x=3.25, y=0.1, size=13, color='green')
ax0.text(s=r'$\omega_\text{max}$', x=1,   y=0.55, size=13, color='green')
ax0.text(s=r'$\omega_\text{min}$', x=-0.9,y=-0.6, size=13, color='green')

# Vertical arrows (keep your positions/colors)
ax0.annotate(text=r'$+\omega$', xy=(3.55, 0.85), xytext=(3.55, 0.45),
             textcoords='data', ha='center', color='blue',
             arrowprops=dict(arrowstyle="->", color="blue", lw=1.5))
ax0.annotate(text=r'$-\omega$', xy=(3.55, -0.85), xytext=(3.55, -0.45),
             textcoords='data', ha='center', color='red',
             arrowprops=dict(arrowstyle="->", color="red", lw=1.5))

# ---------- Panel 2 + physics (consistent SPM model & edges) ----------
from scipy import constants as sc
from scipy.fft import fft, fftfreq, fftshift
from matplotlib.ticker import MultipleLocator, FormatStrFormatter, MaxNLocator

# ----------------- User parameters -----------------
lam0_nm     = 1030.0
t_fwhm_ps   = 1.0
dlam_nm     = 0.1
lam_min_nm  = 900.0
lam_max_nm  = 1150.0

# Medium / pass parameters
L            = 10*33e-3
passes       = 36
pressure_bar = 0.160
n2_Ar_per_bar = 10e-24  # m^2/W per bar
fluence_J_cm2 = 40.0    # time-integrated intensity

# Sampling for FFT simulation
T_window_ps = 200.0
N = 2**18

# ----------------- Derived quantities -----------------
lam0   = lam0_nm * 1e-9
omega0 = 2*np.pi*sc.c/lam0
t_fwhm = t_fwhm_ps * 1e-12

# Intensity Gaussian parameter (I(t) = I0 * exp(-t^2/(2 tau^2)))
tau_s  = t_fwhm/(2*np.sqrt(2*np.log(2)))   # [s]
tau_ps = tau_s * 1e12                      # [ps] for top-axis conversion

T_window = T_window_ps * 1e-12
dt = T_window / N
t_real = (np.arange(N) - N//2) * dt

# Effective n2 and length; peak intensity from fluence (consistent with intensity definition)
n2 = n2_Ar_per_bar * pressure_bar
L_eff = passes * L
fluence_J_m2 = fluence_J_cm2 * 1e4

# Fluence = I0 * sqrt(2π) * tau  -->  I0:
I0 = fluence_J_m2 / (np.sqrt(2*np.pi) * tau_s)

# Field envelope so that |E|^2 ∝ exp(-t^2/(2 tau^2))
A_t = np.exp(-(t_real**2)/(4*tau_s**2))     # amplitude envelope (not intensity)
I_t_phys = np.exp(-(t_real**2)/(2*tau_s**2))

# SPM phase: φ_NL(t) = B * exp(-t^2/(2 tau^2)) with B = k0 n2 L_eff I0
k0 = 2*np.pi/lam0
B_phys = k0 * n2 * L_eff * I0
B = 33.0                                # keep your override; set to B_phys to use physics-based
phi_spm = B * np.exp(-(t_real**2)/(2*tau_s**2))

# Fields
E_t_noSPM = A_t * np.exp(1j*omega0*t_real)
E_t_SPM   = A_t * np.exp(1j*phi_spm) * np.exp(1j*omega0*t_real)

# FFT -> spectrum helper on 900–1150 nm grid
lam_nm_grid = np.arange(lam_min_nm, lam_max_nm + dlam_nm, dlam_nm)
lam = lam_nm_grid * 1e-9
omega_lam = 2*np.pi*sc.c/lam

def spectrum_vs_lambda(E_t):
    f = fftshift(fftfreq(N, d=dt))
    omega = 2*np.pi * f
    E_omega = fftshift(fft(np.fft.ifftshift(E_t))) * dt
    I_omega = np.abs(E_omega)**2
    pos = omega > 0
    omega_pos = omega[pos]
    I_pos = I_omega[pos]
    I_pos /= I_pos.max() if I_pos.max()!=0 else 1.0
    sort_idx = np.argsort(omega_pos)
    I_lambda = np.interp(omega_lam, omega_pos[sort_idx], I_pos[sort_idx], left=0.0, right=0.0)
    return I_lambda, omega_pos[[0,-1]]

I_lambda_no, omega_bounds = spectrum_vs_lambda(E_t_noSPM)
I_lambda_spm, _ = spectrum_vs_lambda(E_t_SPM)

# Plot spectra (keep your colors/limits/legend style)
ax[1].plot(lam_nm_grid, I_lambda_no, label="Before SPM (TL)", color='black')
ax[1].plot(lam_nm_grid, I_lambda_spm, label="After SPM", color='green')
ax[1].set_xlim(1030-38, 1030+38)
ax[1].set_ylim(0, 1.1)
ax[1].set_xlabel("Wavelength (nm)")
ax[1].set_ylabel("Normalized Spectral Intensity (a.u.)")
# ax[1].legend()

# ---------- Correct spectral-edge markers from max instantaneous chirp ----------
# |Δω|_max = B/(τ√e)  (consistent with φ_NL = B exp[-t^2/(2 τ^2)])
d_omega_max = B/(tau_s*np.sqrt(np.e))
r = d_omega_max / omega0
lambda_red  = lam0 / (1 - r)   # longer-λ edge
lambda_blue = lam0 / (1 + r)   # shorter-λ edge

ax[1].axvline(x=lambda_blue*1e9, color='black', ls='dotted', lw=1)
ax[1].axvline(x=lambda_red*1e9,  color='black', ls='dotted', lw=1)
ax[1].text(s=r'$\omega_\text{max}$', x=lambda_blue*1e9-7, y=1.0, size=13, color='black')
ax[1].text(s=r'$\omega_\text{min}$', x=lambda_red*1e9+1, y=1.0, size=13, color='black')
ax[1].text(s=r'$I(\omega)_\text{init}$', x=1032, y=0.78, size=13, color='black')
ax[1].text(s=r'$I(\omega)_\text{SPM}$', x=1010, y=0.82, size=13, color='green')

# ---------- Add requested TOP x-axes (no rescaling of plots) ----------
# Left subplot top axis: time in ps (FWHM = 1 ps -> tau_s as above)
def u_to_tps(u):      # u = t/τ  --> time in ps
    return u * tau_ps
def tps_to_u(tps):    # inverse mapping
    return tps / tau_ps

secax_time = ax[0].secondary_xaxis('top', functions=(u_to_tps, tps_to_u))
secax_time.set_xlabel("Time (ps)")
secax_time.xaxis.set_major_locator(MultipleLocator(0.5))
secax_time.xaxis.set_major_formatter(FormatStrFormatter('%.1f'))

# Right subplot top axis: angular frequency ω in 1e15 rad/s
def lamnm_to_omegap15(lam_nm_val):
    lam_m = lam_nm_val * 1e-9
    return (2*np.pi*sc.c / lam_m) / 1e15
def omegap15_to_lamnm(omega_p15):
    omega = omega_p15 * 1e15
    lam_m = 2*np.pi*sc.c / omega
    return lam_m * 1e9

secax_omega = ax[1].secondary_xaxis('top', functions=(lamnm_to_omegap15, omegap15_to_lamnm))
secax_omega.set_xlabel(r'Angular frequency $\omega$ ($10^{15}$ rad/s)')
secax_omega.xaxis.set_major_locator(MaxNLocator(6))

# ---------- Prints ----------
lam_min_covered = 2*np.pi*sc.c/omega_bounds[1]*1e9
lam_max_covered = 2*np.pi*sc.c/omega_bounds[0]*1e9
print(f"FFT ω coverage maps to roughly λ in [{lam_min_covered:.0f}, {lam_max_covered:.0f}] nm.")
print("Plotted wavelength range: [%.0f, %.0f] nm (step %.1f nm)" % (lam_min_nm, lam_max_nm, dlam_nm))
print("Current SPM peak phase B (override) ≈ %.3f rad; physics-based B ≈ %.3f rad." % (B, B_phys))
print("Predicted max |Δω| = B/(τ√e) ≈ %.3e rad/s  →  |Δω|·τ ≈ %.3f" % (d_omega_max, d_omega_max*tau_s))

# plt.tight_layout()  # keep your choice
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fftshift, fftfreq
from scipy.special import lambertw

# ---------------- Parameters ----------------
tau = 1.0                 # pulse width (arb. units)
B   = 33.0                # nonlinear phase shift
N   = 2**16               # time samples
T_window = 80*tau         # simulation window
t   = np.linspace(-T_window/2, T_window/2, N)
dt  = t[1]-t[0]

# Temporal pulse
I_t = np.exp(-t**2 / (2*tau**2))
E_t = np.sqrt(I_t)

# Nonlinear phase φ(t)
phi = B * np.exp(-t**2 / (2*tau**2))

# Field with SPM
E_t_SPM = E_t * np.exp(1j*phi)

# ---------------- FFT spectrum ----------------
freq = fftshift(fftfreq(N, d=dt))  # frequency grid
Omega_fft = 2*np.pi*freq           # detuning
E_w = fftshift(fft(E_t_SPM)) * dt
I_w = np.abs(E_w)**2

# Normalize FFT at Ω=0
idx0_fft = np.argmin(np.abs(Omega_fft - 0))
I_w /= I_w[idx0_fft]

# ---------------- Stationary-phase envelope ----------------
def stationary_phase_envelope(Omega, B, tau):
    arg = -(Omega**2 * tau**2)/(B**2)
    vals = []
    for branch in [0, -1]:  # two real branches of Lambert W
        Wval = lambertw(arg, k=branch).real
        if Wval < 0:  # valid solutions only
            t_sol = tau * np.sqrt(-Wval)
            for sgn in [+1, -1]:
                t = sgn * t_sol
                I_loc = np.exp(-t**2/(2*tau**2))
                A = B/tau**2
                C = 1/(2*tau**2)
                psi_dd = A*np.exp(-C*t**2)*(2*C*t**2-1)
                if np.abs(psi_dd) > 1e-14:
                    vals.append(I_loc/np.abs(psi_dd))
    return np.sum(vals) if len(vals)>0 else 0.0


# Build Ω grid and evaluate envelope
Omega_grid = np.linspace(-1.5*B/tau, 1.5*B/tau, 2000)
S_env = np.array([stationary_phase_envelope(Om, B, tau) for Om in Omega_grid])

# --- Scale envelope at center ---
idx0_env = np.argmin(np.abs(Omega_grid - 0))
scale_factor = 1.0 / S_env[idx0_env] if S_env[idx0_env] != 0 else 1.0
S_env *= scale_factor

# ---------------- Plot ----------------
dOmega_max = B/(tau*np.sqrt(np.e))

plt.figure(figsize=(8,5))
plt.plot(Omega_fft, I_w, color='black', alpha=0.6, label="FFT spectrum")
plt.plot(Omega_grid, S_env-0.01, color='red', lw=2, label="Stationary-phase envelope (center-normalized)")
plt.xlim(-1.5*dOmega_max, 1.5*dOmega_max)
plt.xlabel(r"Frequency detuning $\Omega$ (arb. units)")
plt.ylabel("Normalized spectral intensity")
plt.legend()
plt.grid(True)
plt.title(f"SPM Spectrum (B={B}, τ={tau})")
plt.ylim(0,2.5)
```

```python
# Revised with controllable label position & angle on the red breakdown contour
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe

# --- Parameters ---
lambda_0 = 1030e-9
M_squared = 1.2
tau_FWHM = 1.3e-12
tau = tau_FWHM/(2*np.sqrt(2*np.log(2)))
E = 80e-3
eta_2 = 10e-24  # m^2/W
f_pick = 4  # picked focal length for the actual build (4 m)

w_small  = 7.7e-3/2
w_large  = 10.0e-3/2
w_design = 8e-3/2

F_lim  = 0.5*1e4  # J/m^2
w_LIDT = np.sqrt(2*E/(np.pi*F_lim))

def B_single(mbar, f, w, both_sides=True):
    p_bar = mbar / 1000.0
    arg = np.pi * w**2 / (M_squared * lambda_0 * f)  # = f/z_R
    pref = (2*np.sqrt(2*np.pi) / (M_squared * lambda_0**2)) * eta_2 * p_bar * (E / tau)
    B = pref * np.arctan(arg)
    return B * (2 if both_sides else 1)

def Fmax_from_f_w(f, w):
    w0 = M_squared*lambda_0*f/(np.pi*w)
    return 2*E/(np.pi*w0**2)  # J/m^2

# --- Grid ---
f_vals = np.linspace(1, 8, 450)        # m
w_vals = np.linspace(2e-3, 8e-3, 360)  # m
F, W = np.meshgrid(f_vals, w_vals)
Wmm = W*1e3
Fmap = Fmax_from_f_w(F, W)

# --- Fluence limits (J/m^2) ---
Flim_160   = 56 * 1e4
Flim_400   = 40 * 1e4
Flim_1000  = 28 * 1e4

# --- User-specified iso-B levels ---
levels_160  = [2.5, 2.6, 2.7, 2.8, 3, 3.3, 3.4, 3.5]
levels_400  = [5, 6, 6.4, 6.6, 6.8, 7, 8, 8.3, 8.5, 8.7]
levels_1000 = [15, 16, 16.5, 17, 17.5, 18, 19, 20, 21, 22, 23]

# --- Compute B maps ---
B160  = B_single(160,  F, W, True)
B400  = B_single(400,  F, W, True)
B1000 = B_single(1000, F, W, True)

# --- Helper: choose point/angle on a contour, with controls ---
def label_on_contour(ax, contour_collection, text, color,
                     x=None, frac=0.5, angle=None, offset_px=14, side='above'):
    """
    Place a label linked to a contour.
      x      : desired x-position in data coords (if None, use frac along arclength)
      frac   : fraction of arclength in [0,1] if x is None
      angle  : rotation [deg]; if None -> align to local tangent
      offset_px : normal offset in *pixels* away from the line
      side   : 'above' or 'below' the line (in +w or -w direction)
    """
    coll = contour_collection.collections[0]
    paths = coll.get_paths()
    if not paths:
        return
    # Use the longest path for robustness
    def _disp_len(p):
        v = p.vertices
        d = ax.transData.transform(v)
        seg = d[1:] - d[:-1]
        return float(np.hypot(seg[:,0], seg[:,1]).sum()) if len(d) > 1 else 0.0
    path = max(paths, key=_disp_len)
    verts = path.vertices
    if len(verts) < 2:
        return

    # Choose index along the path
    if x is not None:
        idx = int(np.argmin(np.abs(verts[:,0] - x)))
    else:
        # arclength-based fraction
        d = ax.transData.transform(verts)
        seg = d[1:] - d[:-1]
        seglen = np.hypot(seg[:,0], seg[:,1])
        cum = np.concatenate([[0], np.cumsum(seglen)])
        L = cum[-1] if len(cum) else 0
        target = np.clip(frac, 0, 1) * (L if L > 0 else 0)
        idx = np.searchsorted(cum, target) - 1
        idx = int(np.clip(idx, 0, len(verts)-2))

    # Local tangent (display coords) around idx
    d = ax.transData.transform(verts)
    i0 = max(0, idx-1); i1 = min(len(verts)-1, idx+1)
    t = d[i1] - d[i0]
    if angle is None:
        angle_deg = float(np.degrees(np.arctan2(t[1], t[0])))
    else:
        angle_deg = float(angle)

    # Base point (display coords) and normal for offset
    pm = 0.5 * (d[idx] + d[idx+1])
    n = np.array([-t[1], t[0]], dtype=float)
    n /= (np.hypot(*n) or 1.0)

    # Ensure "above/below" in data w (mm)
    base_xy  = ax.transData.inverted().transform(pm)
    trial_xy = ax.transData.inverted().transform(pm + 2*n)
    if (side == 'above' and trial_xy[1] < base_xy[1]) or (side == 'below' and trial_xy[1] > base_xy[1]):
        n *= -1

    # Final text position (data)
    x_txt, y_txt = ax.transData.inverted().transform(pm + offset_px*n)
    xmin, xmax = ax.get_xlim(); ymin, ymax = ax.get_ylim()
    m = 0.02*(ymax - ymin)
    x_txt = float(np.clip(x_txt, xmin, xmax))
    y_txt = float(np.clip(y_txt, ymin + m, ymax - m))

    txt = ax.text(x_txt, y_txt, text, color=color, rotation=angle_deg,
                  rotation_mode='anchor', ha='center', va='center',
                  fontsize=13, alpha=0.95, clip_on=True)
    txt.set_path_effects([pe.withStroke(linewidth=3, foreground="white", alpha=0.75)])
    return txt  # so you can keep a handle if you want

# --- Plot ---
fig, axes = plt.subplots(1, 3, figsize=(15,5), sharey=True)

panels = [
    {"ax": axes[0], "B": B160,  "levels": levels_160,  "title": "160 mbar",
     "Flim": Flim_160,  "Flab": r"56 J/cm$^2$",  "label_x": 4, "label_angle": 49},
    {"ax": axes[1], "B": B400,  "levels": levels_400,  "title": "400 mbar",
     "Flim": Flim_400,  "Flab": r"40 J/cm$^2$",  "label_x": 5.7, "label_angle": 46},
    {"ax": axes[2], "B": B1000, "levels": levels_1000, "title": "1000 mbar",
     "Flim": Flim_1000, "Flab": r"28 J/cm$^2$",  "label_x": 6.85, "label_angle": 40},
]

violet = (0.5, 0.0, 0.8)

for p in panels:
    ax = p["ax"]
    # iso-B contours (bigger labels + halo)
    cs = ax.contour(F, Wmm, p["B"], levels=p["levels"], colors='k', linewidths=1.0, alpha=0.7)
    lbls = ax.clabel(cs, fmt='%g', fontsize=11)  # bigger font
    for t in lbls:
        t.set_path_effects([pe.withStroke(linewidth=3, foreground="white", alpha=0.6)])

    # Breakdown region (solid red + red shading above)
    unsafe = np.where(Fmap >= p["Flim"], 1, np.nan)
    ax.contourf(F, Wmm, unsafe, levels=[0.5, 1.5], colors=[(1,0,0,0.08)], zorder=0)
    cF = ax.contour(F, Wmm, Fmap, levels=[p["Flim"]], colors='red', linewidths=2.2, linestyles='solid', zorder=2)

    # NO separate cF label (avoid duplicate) — integrate F_max into the breakdown label:
    breakdown_text = r"optical breakdown ~ " + p["Flab"]
    label_on_contour(ax, cF, breakdown_text, 'red',
                     x=p["label_x"], frac=0.5, angle=p["label_angle"],
                     offset_px=14, side='above')

    # LIDT band (solid violet + violet shading below)
    ax.axhline(y=w_LIDT*1e3, color=violet, linestyle='solid', linewidth=1.6, zorder=3)
    ax.fill_between(f_vals, 0, w_LIDT*1e3, color=violet, alpha=0.08, zorder=0)
    ax.text(0.5*(f_vals.min()+f_vals.max()), w_LIDT*1e3*0.9, 'LIDT',
            ha='center', va='center', color=violet, fontsize=12)

    ax.set_title(p["title"])
    ax.set_xlabel('Focal length f [m]',size=13)
    ax.grid(False, alpha=0.2)
    ax.set_ylim(2.5, 8)
    ax.set_xlim(2, 8)

    # Highlight chosen f and w range/design point
    ax.plot(f_pick*np.ones(20), np.linspace(w_small*1e3, w_large*1e3, 20),
            color='green', lw=3, label='Range of beam radii ')
    ax.scatter(4, w_design*1e3, color='magenta', label='Design choice',s=60,zorder=10)
    ax.legend(loc='upper left',prop={'size': 12})

axes[0].set_ylabel('$1/e^2$ beam radius on optic [mm]',size=13)
#fig.suptitle(f"{round(E*1e3)} mJ, {tau_FWHM*1e12:.1f} ps FWHM, 1030 nm, Argon", fontsize=15)

fig.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------
# Inputs (edit as needed)
# -----------------------
lambda_0 = 1030e-9         # wavelength [m]
F_lim_foc = 56e4           # fluence limit [J/m^2] (e.g., 56 J/cm^2); at 160 mbar in Argon
F_lim_mirror = 0.5e4       # mirror fluence limit [J/m^2]

w_m_min, w_m_max = 3.0e-3, 10.0e-3   # sweep range for mirror 1/e^2 radius [m]
N = 600
w_m = np.linspace(w_m_min, w_m_max, N)

# Scenarios (columns)
E_list = [100e-3, 150e-3, 200e-3]     # [J]
R_list = [4.0, 6.0, 8.0]              # [m]

# Design mirror radius (per column)
w_design_list = [3.67e-3, 4.5e-3, 5.2e-3]

# Top-row (g^2) y-limits, shared
g2_ymin, g2_ymax = 0.92, 1.01

# How far inside to place the right-axis tick labels (points; negative pulls inside)
INSIDE_PAD = -8

# -----------------------
# Helper mappings
# -----------------------
def L_to_g2(L, R):
    """Left axis (L) -> g^2. Using concentric branch relation: g = 1 - L/R."""
    return (1 - L / R) ** 2

def g2_to_L(g2, R):
    """g^2 -> L (right axis), concentric branch: L = (1 + sqrt(g^2)) R."""
    return (1 + np.sqrt(g2)) * R

# -----------------------
# Figure (share x for columns; share y by row)
# -----------------------
fig, ax = plt.subplots(2, 3, figsize=(11, 5), sharex=True, sharey='row')
# We'll tighten spacing at the end with subplots_adjust

for j, (E, R, w_design) in enumerate(zip(E_list, R_list, w_design_list)):
    # -------- Derived (per column) --------
    w0_min = np.sqrt(2 * E / (np.pi * F_lim_foc))      # [m]
    wm_min = np.sqrt(2 * E / (np.pi * F_lim_mirror))   # [m]

    # Cavity relations along the sweep
    kappa = np.pi * w_m**2 / (lambda_0 * R)
    g_star = (1 - kappa**2) / (1 + kappa**2)
    g2_star = g_star**2
    L_star = (1 - g_star) * R
    w0_star = w_m / np.sqrt(1 + kappa**2)

    # Design point
    kappa_d = np.pi * w_design**2 / (lambda_0 * R)
    g_d = (1 - kappa_d**2) / (1 + kappa_d**2)
    g2_design = g_d**2
    L_design = (1 - g_d) * R
    w0_design = w_design / np.sqrt(1 + kappa_d**2)

    # =======================
    # TOP: g^2(w_m) with right axis L (inside)
    # =======================
    ax_top = ax[0, j]

    # Primary left axis: g^2 (shared across all columns)
    if j == 0:
        ax_top.plot(w_m * 1e3, g2_star, lw=2, color='black', label='required mirror separation')
    else:
        ax_top.plot(w_m * 1e3, g2_star, lw=2, color='black')

    ax_top.set_ylim(g2_ymin, g2_ymax)
    if j == 0:
        ax_top.set_ylabel(r'$g^2$',size=12)
    ax_top.grid(alpha=0.2)
    ax_top.hlines(y=1.0,xmin=1,xmax=5.6, color='black', ls='dashed', lw=1)  # L = 2R corresponds to g^2 = 1
    ax_top.text(s=r'$L=2R$', x=3.08, y=1.002, size=11, color='black')

    # Title
    ax_top.set_title('E={:.0f} mJ, R={:.0f} mm'.format(E*1e3, R*1e3))

    # Design point on g^2 curve
    ax_top.scatter(w_design * 1e3, g2_design, s=100,
                   color=('magenta' if j == 0 else 'blue'),
                   edgecolors='white', linewidth=0, zorder=5)

    # Right secondary y-axis: show L scale for THIS column's R
    secax = ax_top.secondary_yaxis(
        'right',
        functions=(lambda g2, R=R: g2_to_L(g2, R),
                   lambda L,  R=R: L_to_g2(L, R))
    )
    secax.set_ylabel('')


    # Keep L ticks consistent with the current g^2 limits:
    secax.set_ylim(g2_to_L(g2_ymin, R), g2_to_L(g2_ymax, R))

    # ---- Move the entire right axis (ticks, ticklabels, label) INSIDE the axes ----
    # Draw tick marks inward and pull tick labels inside with a negative pad
    secax.tick_params(axis='y', which='both', direction='in', right=True,
                      labelright=True, pad=INSIDE_PAD, length=4)
    # Make labels flush toward the right spine (looks cleaner when inside)
    for lbl in secax.get_yticklabels():
        lbl.set_horizontalalignment('right')

    # Place the y-label inside the plot area (slightly left of the right spine)
    # Coordinates are in the secondary axis' axes coordinates (0..1)
    secax.yaxis.set_label_position('right')
    secax.yaxis.set_label_coords(0.7, 0.5)  # tweak 0.97→0.95 to move further inside

    # =======================
    # BOTTOM: w0(w_m) and fluence/mirror constraints
    # =======================
    ax_bot = ax[1, j]
    if j==0:
        ax_bot.plot(w_m * 1e3, w0_star * 1e3, lw=2, color='black',label='supported beam size')
    else:
        ax_bot.plot(w_m * 1e3, w0_star * 1e3, lw=2, color='black')
    ax_bot.set_xlabel(r'$w_m$ [mm]',size=12)
    if j == 0:
        ax_bot.set_ylabel(r'$w_0$ [mm]',size=12)
    ax_bot.grid(alpha=0.2)

    # Fluence-limit line & shading (w0 < w0_min infeasible)
    ax_bot.axhline(w0_min * 1e3, color='red', ls='dashed')
    ymin, ymax = ax_bot.get_ylim()
    ax_bot.fill_between(w_m * 1e3, ymin, w0_min * 1e3, color='red', alpha=0.1)

    # Mirror fluence vertical line & left-fill
    ax_bot.axvline(x=wm_min * 1e3, color='violet', ls='dashed')
    ax_bot.set_xlim(3, 6.2)
    ax_bot.axvspan(3, wm_min * 1e3, color='violet', alpha=0.4)

    # Design point
    if j==0:
        ax_bot.scatter(w_design * 1e3, w0_design * 1e3, s=100, color='magenta',
                    edgecolors='white', linewidth=0, zorder=5)
    else:
        ax_bot.scatter(w_design * 1e3, w0_design * 1e3, s=100, color='blue',
                    edgecolors='white', linewidth=0, zorder=5)

# Unify bottom-row y-limits (same y-axis across all bottom plots)
for j in range(3):
    ax[1, j].set_ylim(0.25, 0.6)

# Legend (only first top subplot has a labeled artist)
ax[0, 0].legend(frameon=False, loc='lower center')
ax[1, 0].legend(frameon=False, loc='upper center')

# Annotations you had
ax[0,0].text(s=' CALA\ndesign', x=3.8, y=0.95, color='magenta')
ax[0,1].text(s='possible\n design', x=4.6, y=0.95, color='blue')
ax[0,2].text(s='possible\n design', x=4.55, y=0.965, color='blue')

ax[1,0].text(s=r'56 J/cm$^2$ (Ar @ 160 mbar)', x=4.3, y=0.35, color='red')
ax[1,1].text(s=r'56 J/cm$^2$', x=5.55, y=0.425, color='red')
ax[1,2].text(s=r'56 J/cm$^2$', x=5.55, y=0.49, color='red')

ax[1,0].text(s=r'LIDT', x=3.4, y=0.53, color='purple',rotation=90,size=11)
ax[1,1].text(s=r'LIDT', x=4.2, y=0.53, color='purple',rotation=90,size=11)
ax[1,2].text(s=r'LIDT', x=4.87, y=0.33, color='purple',rotation=90,size=11)

ax[0,0].text(s='L [m]',x=5.6,y=0.95,rotation=90)
ax[0,1].text(s='L [m]',x=5.55,y=0.95,rotation=90)
ax[0,2].text(s='L [m]',x=5.6,y=0.95,rotation=90)

ax[0,0].text(s='a)',x=3.1,y=0.985,size=15)
ax[0,1].text(s='b)',x=3.1,y=0.985,size=15)
ax[0,2].text(s='c)',x=3.1,y=0.985,size=15)

ax[1,0].text(s='d)',x=3.06,y=0.55,size=15)
ax[1,1].text(s='e)',x=3.06,y=0.55,size=15)
ax[1,2].text(s='f)',x=3.06,y=0.55,size=15)

# ---- Bring subplots closer together ----
plt.subplots_adjust(left=0.07, right=0.985, top=0.95, bottom=0.1,
                    wspace=0.04, hspace=0.22)

plt.savefig('/Users/sebastiankalos/Documents/OXFORD/Thesis_writeup/images/plots/HC_resonator_scaling.png',bbox_inches='tight',dpi=300)
```

```python
# Final refinements:
# - Red curvature lines now *fade with the Gaussian*: alpha at each segment ∝ exp(-2 y^2 / w(z)^2).
#   (Implemented with a LineCollection for per-segment alpha.)
# - Mirrors thicker and labeled **M1**, **M2**.
# - Remove z_R tick labels to declutter.
# - Waist annotation now shows the **radius** (1/e^2) with a single upward arrow from the axis to +w0,
#   made a bit larger and centered.
# - Keep strict curved clipping by the mirrors, gentler axial falloff for the red fill, and 13 lines with linear z spacing.
#
# Tip: adjust LINE_ALPHA_SCALE to tune the visibility of the fading on the curvature lines.

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.collections import LineCollection

# ----- Parameters -----
R = 10.0
L = 19.0
lam = 1.0
assert L < 2*R

# Beam parameters
z_R = (L/2) * np.sqrt((2*R/L) - 1.0)
w0 = np.sqrt(z_R * lam / np.pi)

def w(z):
    return w0 * np.sqrt(1.0 + (z/z_R)**2)

# Mirror circle centers
cx_left  = -L/2 + R
cx_right =  L/2 - R

# Visualization tuning
N_TOTAL = 13
P_AXIAL = 1.0
EPS_FRAC = 0.05        # margin from mirrors
EPS = EPS_FRAC * L
ARC_WINDOW = 1.2       # radians half-window for arcs around axis-crossing
MIRROR_LW = 4.5        # thicker mirrors
LINE_ALPHA_SCALE = 0.95  # overall scale for curvature-line alpha

# ----- Figure -----
fig, ax = plt.subplots(figsize=(10, 4))

# ROC circles (dashed)
theta = np.linspace(0, 2*np.pi, 2400)
x_left_circle  = cx_left  + R*np.cos(theta)
y_left_circle  = 0        + R*np.sin(theta)
x_right_circle = cx_right + R*np.cos(theta)
y_right_circle = 0        + R*np.sin(theta)
ax.plot(x_left_circle,  y_left_circle,  linestyle='dotted', color='black', linewidth=1, label='Mirror ROC')
ax.plot(x_right_circle, y_right_circle, linestyle='dotted', color='black', linewidth=1)

# Mirror surface arcs (thicker, black)
arc_span_deg = 35
phi = np.deg2rad(np.linspace(-arc_span_deg, arc_span_deg, 600))
x_left_arc = cx_left + R*np.cos(np.pi + phi)
y_left_arc = 0       + R*np.sin(np.pi + phi)
x_right_arc = cx_right + R*np.cos(phi)
y_right_arc = 0        + R*np.sin(phi)
ax.plot(x_left_arc, y_left_arc, color='black', linewidth=MIRROR_LW)
ax.plot(x_right_arc, y_right_arc, color='black', linewidth=MIRROR_LW)

# ----- Red Gaussian fill (strict curved clipping), gentle axial falloff -----
zmin, zmax = -L/2, L/2
yspan = 0.75 * R
Ny, Nz = 600, 2400
yg = np.linspace(-yspan, yspan, Ny)
zg = np.linspace(zmin, zmax, Nz)
Y, Z = np.meshgrid(yg, zg, indexing='ij')

W = w(Z)
G_y = np.exp(-2.0 * (Y**2) / (W**2))
I_peak = (w0 / W)**P_AXIAL
I = I_peak * G_y
I_norm = I / np.max(I)

inside_left  = (Z - cx_left )**2 + Y**2 <= R**2
inside_right = (Z - cx_right)**2 + Y**2 <= R**2
inside_cavity = inside_left & inside_right

rgba = np.zeros((Ny, Nz, 4), dtype=float)
rgba[..., 0] = 1.0
rgba[..., 1] = 0.0
rgba[..., 2] = 0.0
rgba[..., 3] = 0.5 * I_norm * inside_cavity
ax.imshow(rgba, extent=[zmin, zmax, -yspan, yspan], origin='lower', aspect='auto')

# ----- Curvature lines: linear z positions away from mirrors; fade with Gaussian in alpha -----
def Rmag(z):
    z = np.asarray(z, dtype=float)
    Rm = np.empty_like(z)
    near0 = np.isclose(z, 0.0)
    Rm[near0] = 1e12
    znz = z[~near0]
    Rm[~near0] = np.abs(znz) * (1.0 + (z_R**2)/(znz**2))
    return Rm

def wavefront_center(z0):
    s = np.sign(z0) if z0 != 0 else 1.0
    return z0 - s * Rmag(z0)

z_positions = np.linspace(zmin + EPS, zmax - EPS, N_TOTAL)

for z0 in z_positions:
    # Require axis point inside both mirror disks
    if not (((z0 - cx_left)**2 <= R**2) and ((z0 - cx_right)**2 <= R**2)):
        continue

    if np.isclose(z0, 0.0, atol=1e-12):
        # Waist: vertical line at z=0, use LineCollection with alpha varying by y
        y_vals = np.linspace(-yspan, yspan, 1600)
        mask_circles = ((0 - cx_left )**2 + y_vals**2 <= R**2) & ((0 - cx_right)**2 + y_vals**2 <= R**2)
        mask_env = np.abs(y_vals) <= w0
        mask = mask_circles & mask_env
        y_line = y_vals[mask]
        if len(y_line) > 1:
            # build segments in y with alpha from Gaussian
            pts = np.column_stack((np.zeros_like(y_line), y_line))
            segs = np.stack([pts[:-1], pts[1:]], axis=1)
            # alpha per segment
            wloc = w(0.0)
            alpha_seg = LINE_ALPHA_SCALE * np.exp(-2.0 * (0.5*(y_line[:-1] + y_line[1:]))**2 / (wloc**2))
            colors = np.tile([1.0, 0.0, 0.0, 1.0], (len(segs), 1))
            colors[:, 3] = alpha_seg
            lc = LineCollection(segs, colors=colors, linewidths=1.1)
            ax.add_collection(lc)
        continue

    Rc = Rmag(z0)
    cx = wavefront_center(z0)

    cos_target = np.clip((z0 - cx)/Rc, -1.0, 1.0)
    theta_target = np.arccos(cos_target)

    thetas = np.linspace(theta_target - ARC_WINDOW, theta_target + ARC_WINDOW, 3000)
    thetas = np.concatenate([thetas, -thetas])

    xw = cx + Rc*np.cos(thetas)
    yw = 0  + Rc*np.sin(thetas)

    # Strict mirror & envelope clipping
    inside_left_curve  = (xw - cx_left )**2 + yw**2 <= R**2
    inside_right_curve = (xw - cx_right)**2 + yw**2 <= R**2
    mask_cavity = inside_left_curve & inside_right_curve
    if not np.any(mask_cavity):
        continue

    x_in = xw[mask_cavity]
    y_in = yw[mask_cavity]

    w_in = w(x_in)
    mask_env = np.abs(y_in) <= w_in
    if not np.any(mask_env):
        continue

    xi = x_in[mask_env]
    yi = y_in[mask_env]

    # Build segments and apply per-segment alpha based on Gaussian at (xi, yi)
    if len(xi) > 1:
        pts = np.column_stack((xi, yi))
        segs = np.stack([pts[:-1], pts[1:]], axis=1)
        # alpha per segment using average y and local w(x)
        wloc = 0.5*(w(xi[:-1]) + w(xi[1:]))
        ymid = 0.5*(yi[:-1] + yi[1:])
        alpha_seg = LINE_ALPHA_SCALE * np.exp(-2.0 * (ymid**2) / (wloc**2))
        colors = np.tile([1.0, 0.0, 0.0, 1.0], (len(segs), 1))
        colors[:, 3] = alpha_seg
        lc = LineCollection(segs, colors=colors, linewidths=1.1)
        ax.add_collection(lc)

# ----- Annotate waist w0 as a radius (single upward arrow from axis) -----
ax.annotate("", xy=(0, w0*1.1), xytext=(0, 0),
            arrowprops=dict(arrowstyle="->", linewidth=1.6, color='black'))
ax.text(0.0*R, w0*1.1, r"$w_0$", va='bottom', ha='left',size=12)

# ----- Lower-side radius arrows (kept subtle) -----
diag = np.deg2rad(35.0)
end_left_low = (cx_left + R*np.cos(-np.pi + diag), R*np.sin(-np.pi + diag))
ax.annotate("", xy=end_left_low, xytext=(cx_left, 0),
            arrowprops=dict(arrowstyle="->", linewidth=1.0, color='black', alpha=0.7,ls='dashed'))
ax.text((cx_left + end_left_low[0])/2 - 0.2*R, (0 + end_left_low[1])/2 - 0.05*R, "R",
        ha='right', va='top', color='black', alpha=0.7)

end_right_low = (cx_right + R*np.cos(-diag), R*np.sin(-diag))
ax.annotate("", xy=end_right_low, xytext=(cx_right, 0),
            arrowprops=dict(arrowstyle="->", linewidth=1.0, color='black', alpha=0.7,ls='dashed'))
ax.text((cx_right + end_right_low[0])/2 + 0.2*R, (0 + end_right_low[1])/2 - 0.05*R, "R",
        ha='left', va='top', color='black', alpha=0.7)

# ----- Dimension arrow L (unchanged) -----
y_annot = -0 * yspan
ax.annotate("", xy=(zmin, y_annot), xytext=(zmax, y_annot),
            arrowprops=dict(arrowstyle="<->", linewidth=1.0, color='black', alpha=0.7,linestyle='dashed'))
ax.text(-5, 0.8, "L", ha='center', va='top', color='black')

# ----- Final formatting -----
ax.text(
    0.02, 0.9,                                  # x,y in axes fraction (0..1)
    "L < 2R\n"
    "g = 1 - L/R = {:.1f}\n"
    "g² = {:.2f}".format(1 - L/R, (1 - L/R)**2),
    transform=ax.transAxes,                      # use axes coords
    ha='left', va='center',                      # left-align the block
    multialignment='left',                       # left-align each line
    fontsize=11, color='black',
    linespacing=1.25
    # , bbox=dict(facecolor='white', alpha=0.6, edgecolor='none', pad=2)  # optional
)


ax.set_aspect('equal', adjustable='box')
ax.set_xlim(-L/2 - 0.6*R, L/2 + 0.6*R)
ax.set_ylim(-yspan, yspan)
ax.set_xlabel("Optical axis (z) [a.u.]")
ax.set_ylabel("Transverse coordinate (r) [a.u.]")

# Legend minimal
handles, labels = ax.get_legend_handles_labels()
unique = dict(zip(labels, handles))
if unique:
    ax.legend(unique.values(), unique.keys(), loc="upper right", frameon=False)

ax.text(-7.5, 4.5, "M1", ha='center', va='top', fontsize=12, color='black')
ax.text(7.5, 4.5, "M2", ha='center', va='top', fontsize=12, color='black')

plt.tight_layout()
```
