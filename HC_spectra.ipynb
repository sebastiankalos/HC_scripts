```python
import matplotlib.pyplot as plt
import numpy as np

def plot_data_from_file(path, filename, label,color,plot_title="Data Plot"):
    """
    Reads data from a text file and plots the wavelength vs intensity.
    
    Parameters:
    filename (str): The name of the text file to read data from.
    plot_title (str): The title of the plot.
    """
    # Initialize lists to hold the wavelength and intensity data
    wavelengths = []
    intensities = []

    filename = path + filename
    
    # Open the file and read the data starting from the 15th line
    with open(filename, 'r') as file:
        lines = file.readlines()[14:]  # Skip the first 14 lines
        
        for line in lines:
            if line.strip():  # Skip empty lines
                parts = line.split()
                wavelengths.append(float(parts[0]))
                intensities.append(float(parts[1]))

    intensities = np.array(intensities)/np.max(intensities)
    
    # Plot the data
    plt.plot(wavelengths, intensities, label=label,color=color)
    plt.xlabel('Wavelength [nm]')
    plt.ylabel('Intensity [arb.units]')
    plt.title(plot_title)
    plt.xlim(980,1080)
    plt.legend()
    plt.grid(True,alpha=0.5)


path = '/Users/sebastiankalos/Documents/OXFORD/HerriottCell/HC_test_240704/'
```
